````python
import copy

a = [[1,2,3]]
# list가 2차원일때 deep copy을 하기 위해서 copy모듈에 있는 
# deepcopy메소드를 사용한다.
b = copy.deepcopy(a) 

print(a[0] is b[0])        # 메모리 주소 False
print(a[0] == b[0])        # 값 True

b[0][0] = 10

print(a[0] is b[0])         # 메모리 주소 False
print(a[0] == b[0])         # 값 False

print(a)                    # [[1, 2, 3]]
print(b)                    # [[10, 2, 3]]
````

````python
import numpy as np

# 1. Shape (4,) 크기의 1차원 배열 생성
arr = np.array([1, 2, 3, 4])

# 2. 스칼라(단일 값) 정의
scalar = 1

# 3. 배열과 스칼라 간의 덧셈 연산 (브로드캐스팅 발생)
#    스칼라 값 '1'이 배열의 크기인 (4,)에 맞게 [1, 1, 1, 1]로 확장되어 각 원소별로 더해집니다.
result = arr + scalar

# ==============================================================================
# 출력 및 결과
# ==============================================================================
print(arr)         # 입력 배열 출력 -> 출력: [1 2 3 4]
print(scalar)      # 더해지는 스칼라 값 출력 -> 출력: 1
print(result)      # 연산 결과 출력 -> 출력: [2 3 4 5]

# Shape(모양) 확인
print(arr.shape)   # 입력 배열의 shape -> 출력: (4,)
print(result.shape)# 결과 배열의 shape -> 출력: (4,)
````

````python
import numpy as np

# 1. 2차원 배열 x2 생성 (shape: 3x3)
x2 = np.array([[2, 1, 6],
               [0, 7, 4],
               [5, 3, 2]])

# 2. 정렬 연산
# ① 좌에서 우로 정렬 (axis=1: 행 내부에서 가로 방향 정렬)
sort_axis1 = np.sort(x2, axis=1)

# ② 위에서 아래로 정렬 (axis=0: 열 내부에서 세로 방향 정렬)
sort_axis0 = np.sort(x2, axis=0)

# ③ 아래에서 위로 거꾸로 정렬 (axis=0 정렬 후 행 순서 뒤집기)
sort_axis0_reverse = np.sort(x2, axis=0)[::-1]

# ==============================================================================
# 출력 및 결과
# ==============================================================================
print(x2)
# 원본 배열 출력:
# [[2 1 6]
#  [0 7 4]
#  [5 3 2]]

# ① np.sort(x2, axis=1) -> 좌에서 우로 정렬
print(sort_axis1)
# 출력:
# [[1 2 6]
#  [0 4 7]
#  [2 3 5]]

# ② np.sort(x2, axis=0) -> 위에서 아래로 정렬
print(sort_axis0)
# 출력:
# [[0 1 2]
#  [2 3 4]
#  [5 7 6]]

# ③ np.sort(x2, axis=0)[::-1] -> 아래에서 위로 거꾸로 정렬
print(sort_axis0_reverse)
# 출력:
# [[5 7 6]
#  [2 3 4]
#  [0 1 2]]
````

````python
import numpy as np

# 1. 1차원 배열 생성 (색인/Index: 0~5)
#    값(Value)   : [5, 4, 3, 2, 1, 0]
#    인덱스(Index): [0, 1, 2, 3, 4, 5]
x = np.array([5, 4, 3, 2, 1, 0])

# 2. 통계 및 조건부 추출 연산
max_val = np.max(x)                 # 최대값
min_val = np.min(x)                 # 최소값
where_idx = np.where(x >= 3)        # 조건(x >= 3)을 만족하는 인덱스 추출
where_val = x[np.where(x >= 3)]     # 조건(x >= 3)에 해당하는 실제 값들 추출

# ==============================================================================
# 출력 및 결과
# ==============================================================================
print(x)                  # 입력 배열 -> 출력: [5 4 3 2 1 0]

# 1) 최대값 (np.max)
print(max_val)            # 출력: 5

# 2) 최소값 (np.min)
print(min_val)            # 출력: 0

# 3) 조건에 해당하는 인덱스 위치 (np.where(x >= 3))
#    x >= 3 조건(5, 4, 3)을 만족하는 위치의 인덱스 [0, 1, 2] 반환
print(where_idx)          # 출력: (array([0, 1, 2]),)

# 4) 조건에 해당하는 실제 값 (x[np.where(x >= 3)])
#    인덱스 [0, 1, 2]에 위치한 실제 요소들을 추출
print(where_val)          # 출력: [5 4 3]
````

````python
 
import numpy as np

# 1. 벡터 및 행렬 생성
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])

A = np.array([[1, 2], 
              [3, 4]])

B = np.array([[5, 6], 
              [7, 8]])

# 2. 벡터 내적 (Dot Product)
dot_product = np.dot(v1, v2)
print("Vector Dot Product:")
print(dot_product)
# 출력: 32 (1*4 + 2*5 + 3*6)

# 3. 행렬 곱셈 (Matrix Multiplication)
matrix_mul = np.matmul(A, B) # 또는 A @ B
print("\nMatrix Multiplication (A @ B):")
print(matrix_mul)
# 출력:
# [[19 22]
#  [43 50]]

# 4. 전치 행렬 (Transpose Matrix)
A_transpose = A.T
print("\nTranspose of Matrix A:")
print(A_transpose)
# 출력:
# [[1 3]
#  [2 4]]

# 5. 역행렬 (Inverse Matrix)
A_inv = np.linalg.inv(A)
print("\nInverse of Matrix A:")
print(A_inv)
# 출력:
# [[-2.   1. ]
#  [ 1.5 -0.5]]

# 6. 행렬식 (Determinant)
det_A = np.linalg.det(A)
print("\nDeterminant of Matrix A:")
print(det_A)
# 출력: -2.0 (1*4 - 2*3)

# 7. 고유값 및 고유벡터 (Eigenvalues & Eigenvectors)
eigenvalues, eigenvectors = np.linalg.eig(A)
print("\nEigenvalues of A:")
print(eigenvalues)
# 출력: [-0.37228132  5.37228132]

print("\nEigenvectors of A:")
print(eigenvectors)
# 출력:
# [[-0.82456484 -0.41597356]
#  [ 0.56576746 -0.90937671]]
````

````python
import numpy as np

# 1. 시드 고정 및 가상의 학생 성적 데이터셋 생성 (학생 5명, 과목 3개)
np.random.seed(42)
scores = np.random.randint(40, 101, size=(5, 3))

print("1. 원본 성적 데이터셋 (5명 x 3과목):")
print(scores)

# 2. 각 과목별 평균 및 표준편차 구하기 (axis=0: 세로/열 방향 연산)
subject_mean = np.mean(scores, axis=0)
subject_std = np.std(scores, axis=0)

print("\n2. 과목별 평균:")
print(subject_mean)

print("\n3. 과목별 표준편차:")
print(np.round(subject_std, 2))

# 3. 각 학생별 평균 구하기 (axis=1: 가로/행 방향 연산)
student_mean = np.mean(scores, axis=1)

print("\n4. 학생별 평균 점수:")
print(np.round(student_mean, 2))

# 4. 낙제점 수정하기 (60점 미만인 점수를 60점으로 일괄 수정)
# np.where(조건, 참일 때 값, 거짓일 때 값) 활용
fixed_scores = np.where(scores < 60, 60, scores)

print("\n5. 낙제점(60점 미만) 수정 후 데이터셋:")
print(fixed_scores)
````

````python
import numpy as np

# 1. 가상의 RGB 이미지 픽셀 데이터 생성 (높이 4 x 너비 4 x 채널 3)
np.random.seed(100)
image_rgb = np.random.randint(0, 256, size=(4, 4, 3), dtype=np.uint8)

print("1. 원본 RGB 이미지 Shape:")
print(image_rgb.shape)

# 2. RGB 흑백 전환 (Grayscale Conversion)
# ITU-R BT.601 표준 공식: Y = 0.299*R + 0.587*G + 0.114*B
# 채널 축(axis 2)에 가중치를 내적(Dot Product)하거나 브로드캐스팅하여 계산함
weights = np.array([0.299, 0.587, 0.114])
image_gray = np.dot(image_rgb, weights).astype(np.uint8)

print("\n2. 흑백 전환 후 이미지 Shape:")
print(image_gray.shape)  # (4, 4)로 차원 축소됨

print("\n3. 흑백 이미지 픽셀 값:")
print(image_gray)

# 3. 이미지 크기 조절 (Downsampling / Resizing)
# 슬라이싱([::2, ::2])을 이용하여 가로/세로 픽셀을 2칸씩 스킵함으로써 크기를 1/2로 줄임
image_resized = image_gray[::2, ::2]

print("\n4. 크기 조절(2x2로 축소) 후 이미지 Shape:")
print(image_resized.shape)

print("\n5. 최종 축소된 이미지 픽셀 값:")
print(image_resized)
````
